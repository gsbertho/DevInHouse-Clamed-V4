### [M1S05] Ex. 01 - Leitura e Estruturação dos Dados (Pandas)

Importe os arquivos clientes.csv, pedidos.csv e produtos.csv usando Pandas. Explore os dados para entender sua estrutura.

```python
#import da biblioteca
import pandas as pd #Onde pd é o alias (apelido)
```

```python
#Leitura e armazenamento dos arquivos em DataFrames

df_clientes = pd.read_csv('clientes.csv')
df_pedidos = pd.read_csv('pedidos.csv')
df_produtos = pd.read_csv('produtos.csv')

#explorar as primeiras linhas de cada df
dfs_ativos = [
    ('clientes',df_clientes),
    ('pedidos',df_pedidos),
    ('produtos',df_produtos)
]

for nome, df in dfs_ativos:
    print(f"Primeiras 4 linhas do df {nome}:")
    print(df.head(4))
    print('\n')


#verificar tipos e nulos
for nome, df in dfs_ativos:
    print(f"Verificação de tipo e nulos do df {nome}")
    print(df.info())
    print(df.isnull().sum())
    print('\n')

#obter estatísticas de cada df
for nome, df in dfs_ativos:
    print(f"Estatísticas base do df {nome}")
    print(df.describe())
    print('\n')
    
```

### [M1S05] Ex. 02 - Filtros e Seleções
Aplique filtros para extrair informações específicas dos DataFrames.

```python
#filtrar dfs de acordo com as especificações
df_clientes_filtrado = df_clientes[df_clientes['idade'] > 30]
df_pedidos_filtrado = df_pedidos[df_pedidos['valor_total']>500]
df_produtos_filtrado = df_produtos[df_produtos['categoria']=='Eletrônicos']

#exemplos selecoes com iloc e loc

print('selecionar coluna inteira de data de df_pedidos_filtrado')
print(df_pedidos_filtrado.loc[:,'data_pedido'])
print('\n')

print('selecionar segunda linha inteira em df_produtos_filtrado')
print(df_produtos_filtrado.iloc[1,:])

```

### [M1S05] Ex. 03 - Operações com Dataframes
Realize operações para cruzar e resumir os dados.

```python
#juntar clientes e pedidos
df_merged = df_pedidos.merge(df_clientes,on='id_cliente')

#agrupar por clientes e somar valores
df_agrupado = df_merged.groupby(['id_cliente','nome'])['valor_total'].sum()
df_agrupado
```

```python
#exibir df_produtos para facilitar o proximo passo
df_produtos
```

```python
#criar tabela dinamica com categorias e medias de preços
tabdim = pd.pivot_table(df_produtos,values='preco',index='categoria',aggfunc='mean',margins=True)
tabdim
```

```python
#verificar duplicatas
df_pedidos.duplicated()
```

```python
#limpar duplicatas (caso fosse necessario)
df_pedidos = df_pedidos.drop_duplicates() 
```

### [M1S05] Ex. 04 - Análise Numérica com NumPy

Utilize NumPy para cálculos e simulações com arrays.


```python
#importar numpy
import numpy as np

#criar array com valores dos pedidos
array_pedidos = np.array(df_pedidos['valor_total'])

print("Array criado: ",array_pedidos,'\n')

#calcular média, soma, desvio padrão
print('Média: ',array_pedidos.mean())
print('Soma: ',array_pedidos.sum())
print('Desvio Padrão',array_pedidos.std())
print('\n')

#Aplicar Desconto de 10% com broadcasting
array_pedidos_com_desconto = array_pedidos * 0.9
print("Array com desconto de 10%: ",array_pedidos_com_desconto)
print('\n')

#Verificar tipos e dimensões com .dtype() e .shape()
print('Dtype: ',array_pedidos.dtype)
print('Shape: ',array_pedidos.shape)
```

### [M1S05] Ex. 05 - Relatório Final
Produza um resumo com os principais insights da análise.

```python
#Identificar cliente que mais gastou
print('Cliente que mais gastou: ',
     df_merged.groupby('nome')['valor_total'].sum().sort_values(ascending=False).index[0],
     df_merged.groupby('nome')['valor_total'].sum().sort_values(ascending=False).iloc[0]
     )
print('\n')

#Calcular ticket médio por categoria
print("Ticket Médio por categoria: ")
print("Analisado que não é possível obter esse dado pois não temos dados de vendas abertas por produto")
print('\n')

#Verificar cidade com maior número de clientes
print("Cidade com maior número de clientes: " )
print("Clientes por cidade: ")
print(df_clientes.groupby('cidade')['id_cliente'].count())
print("Todas cidades com um cliente cada")

#Criar Visualizações (Opcional)
#importar pyplot
import matplotlib.pyplot as plt

#Gráfico de barras de gasto por cliente
ax = df_merged.groupby('nome')['valor_total'].sum().sort_values().plot(kind='bar')
plt.title('Gasto por cliente')
plt.xlabel('Nome Cliente')
plt.ylabel('Gasto em reais')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.show()
```
